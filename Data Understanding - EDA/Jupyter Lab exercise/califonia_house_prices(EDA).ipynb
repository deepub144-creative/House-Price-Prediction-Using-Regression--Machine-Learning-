```python
# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")


# 2. DISPLAY SETTINGS (SHOW ALL ROWS & COLUMNS)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)


# 3. LOAD DATASET
df = pd.read_csv(r"C:/Users/deepu/Downloads/housing.csv")

print("FULL DATASET:")
print(df)  

# 4. BASIC INFORMATION
print("\nShape:", df.shape)
print("\nColumns:\n", df.columns)

print("\nInfo:")
df.info()

print("\nSummary Statistics:")
print(df.describe())


# 5. MISSING VALUES
print("\nMissing Values:\n", df.isnull().sum())


# 6. TARGET VARIABLE ANALYSIS
plt.figure()
sns.histplot(df["median_house_value"], kde=True)
plt.title("Distribution of House Prices")
plt.show()

# 7. NUMERICAL DISTRIBUTIONS
df.hist(bins=30, figsize=(12,10))
plt.suptitle("Feature Distributions")
plt.show()


# 8. CORRELATION HEATMAP
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()


# 9. SCATTER PLOTS
sns.scatterplot(x="median_income", y="median_house_value", data=df)
plt.title("Income vs Price")
plt.show()

sns.scatterplot(x="total_rooms", y="median_house_value", data=df)
plt.title("Rooms vs Price")
plt.show()


# 10. OUTLIER DETECTION
sns.boxplot(x=df["median_house_value"])
plt.title("Outliers in Price")
plt.show()


# 11. CATEGORICAL ANALYSIS
sns.countplot(x="ocean_proximity", data=df)
plt.xticks(rotation=45)
plt.title("Ocean Proximity Count")
plt.show()

sns.boxplot(x="ocean_proximity", y="median_house_value", data=df)
plt.xticks(rotation=45)
plt.title("Price by Location")
plt.show()


# 12. PAIRPLOT 
sns.pairplot(df.sample(1000))
plt.show()


# 13. SKEWNESS
print("\nSkewness:\n", df.skew(numeric_only=True))


# 14. RESET DISPLAY SETTINGS 
pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')

print("\nEDA COMPLETED")
```
